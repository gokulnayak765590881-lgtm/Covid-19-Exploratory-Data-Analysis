#7. Data Aggregation:
 a. Find max value in all columns using groupby function on
'continent'
 column

 Tip: use reset_index() after applying groupby

 b. Store the result in a new dataframe named 'df_groupby'.
 (Use df_groupby dataframe for all further analysis)

```python
# Group by continent and find maximum of all columns
df_groupby = df.groupby('continent').max().reset_index()

# Display result
print(df_groupby.head())
```

```python
# Group by continent and get maximum values
df_groupby = df.groupby('continent').max().reset_index()

# View the new dataframe
print(df_groupby.head())
```

#8. Feature Engineering :
 a. Create a new feature 'total_deaths_to_total_cases' by ratio of
 'total_death

```python
# Create new ratio feature
df_groupby['total_deaths_to_total_cases'] = (
    df_groupby['total_deaths'] / df_groupby['total_cases']
)

# View result
print(df_groupby[['continent',
                  'total_cases',
                  'total_deaths',
                  'total_deaths_to_total_cases']])
```

```python
# Create new ratio feature
df_groupby['total_deaths_to_total_cases'] = (
    df_groupby['total_deaths'] / df_groupby['total_cases'].replace(0, 1)
)
# View result
print(df_groupby[['continent','total_cases','total_deaths','total_deaths_to_total_cases']])
```

#9. Data Visualization :

```python
# Univariate analysis using distplot
sns.distplot(df_groupby['gdp_per_capita'])

plt.title("Histogram of GDP per Capita")
plt.xlabel("GDP per Capita")
plt.ylabel("Density")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='gdp_per_capita',
                y='total_cases',
                data=df_groupby)

plt.title("Total Cases vs GDP per Capita")
plt.xlabel("GDP per Capita")
plt.ylabel("Total Cases")
plt.show()

# Total_cases and the GDP_Per_capital
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='gdp_per_capita',
                y='total_cases',
                hue='continent',
                data=df_groupby)
plt.show()
```

```python
sns.catplot(x='continent',
            y='total_cases',
            data=df_groupby,
            kind='bar')

plt.title("Total Cases by Continent")
plt.xticks(rotation=45)
plt.show()
```
